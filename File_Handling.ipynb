```python
import os
import sys
import random
```

```python

def process_folder(folder_path: str, num_files: int = 10) -> int:
    """
    Process a folder by creating files, checking contents, 
    deleting half randomly, and verifying the result.

    :param folder_path: The path to the folder to process.
    :type folder_path: str

    :param num_files: Number of files to create in the folder. Default is 10.
    :type num_files: int

    :raise ValueError: If the folder path exists but is not a directory.
    :raise FileExistsError: If a file with the same name as the folder exists.

    :return: Number of files remaining after deletion.
    :rtype: int
    """

    # Check if path exists and is not a directory
    if os.path.exists(folder_path):
        if not os.path.isdir(folder_path):
            raise ValueError("The given path exists but is not a directory.")
    else:
        # If path doesn't exist, create the directory
        os.makedirs(folder_path)

    # Create 'num_files' in the folder
    for i in range(num_files):
        file_path = os.path.join(folder_path, f"file_{i+1}.txt")
        with open(file_path, 'w') as f:
            f.write(f"This is file number {i+1}.")

    # Get all files in folder
    files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    original_len = len(files)

    # Delete half of them randomly
    num_to_delete = original_len // 2
    files_to_delete = random.sample(files, num_to_delete)
    for file_name in files_to_delete:
        os.remove(os.path.join(folder_path, file_name))

    # Get remaining files
    remaining_files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

    # Return a summary (optional: could also return list of remaining files)
    return {
        "original_file_count": original_len,
        "deleted_files": len(files_to_delete),
        "remaining_files": len(remaining_files),
        "remaining_file_names": remaining_files
    }
```
