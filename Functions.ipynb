```python
# This function takes a number from the user and prints its multiplication table from 1 to 10

def print_multiplication_table(n):
    print(f"\nMultiplication table of {n}:")
    for i in range(1, 11):
        print(f"{n} × {i} = {n * i}")

def main_multiplication_table():
    try:
        number = float(input("Enter a number to print its multiplication table: "))
        print_multiplication_table(number)
    except ValueError:
        print("Please enter a valid integer!")

main_multiplication_table()
```

```python
# This code prints all twin primes less than 1000 (two prime numbers that differ by 2)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def find_twin_primes(limit):
    print("\nTwin Primes less than", limit, ":")
    for i in range(3, limit, 2):
        if is_prime(i) and is_prime(i + 2):
            print(f"Twin prime: {i} and {i+2}")

def main_twin_primes():
    find_twin_primes(1000)

main_twin_primes()
```

```python
# This function checks whether a number is prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# This function returns the prime factors of a number as a list
def prime_factors(n):
    factors = []
    i = 2
    while i <= n:
        if n % i == 0:
            factors.append(i)
            n = n // i
        else:
            i += 1
    return factors

# This is the main function that handles user input and displays results
def main_prime_factors():
    try:
        num = int(input("\nEnter a number to find its prime factors: "))
        
        if num <= 1:
            print("Please enter a number greater than 1.")
        elif is_prime(num):
            print(f"{num} is a prime number, so it has no prime factors other than itself.")
        else:
            factors = prime_factors(num)
            print("Prime factors:", factors)
    except ValueError:
        print("Invalid input! Please enter an integer.")

main_prime_factors()
```

```python
# This function converts a decimal number (including fractions) to binary manually

def convert_integer_to_binary(integer_part):
    if integer_part == 0:
        return "0"
    binary = ""
    while integer_part > 0:
        binary = str(integer_part % 2) + binary
        integer_part //= 2
    return binary

def convert_fraction_to_binary(fraction_part, precision=10):
    binary_fraction = ""
    count = 0
    while fraction_part > 0 and count < precision:
        fraction_part *= 2
        bit = int(fraction_part)
        binary_fraction += str(bit)
        fraction_part -= bit
        count += 1
    return binary_fraction or "0"

def dec_to_bin(number):
    integer_part = int(number)
    fraction_part = number - integer_part

    integer_binary = convert_integer_to_binary(integer_part)
    fraction_binary = convert_fraction_to_binary(fraction_part)

    if fraction_part == 0:
        return integer_binary
    else:
        return f"{integer_binary}.{fraction_binary}"

def main_decimal_to_binary():
    try:
        number = float(input("\nEnter a decimal number to convert to binary: "))
        print(f"Binary of {number} is: {dec_to_bin(number)}")
    except ValueError:
        print("Please enter a valid decimal number.")

main_decimal_to_binary()
```

```python
# This code prints all perfect numbers in a user-defined range (a number equal to the sum of its proper divisors)

def is_perfect(n):
    if n < 2:
        return False
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    return total == n

def perfect_numbers_in_range(start, end):
    perfects = []
    for num in range(start, end + 1):
        if is_perfect(num):
            perfects.append(num)
    return perfects

def main_perfect_numbers():
    try:
        start = int(input("\nEnter start of range: "))
        end = int(input("Enter end of range: "))
        print("Perfect numbers in range:", perfect_numbers_in_range(start, end))
    except ValueError:
        print("Please enter valid integers.")

main_perfect_numbers()
```
